```python
import asyncio
import pandas as pd
from ragas.llms import llm_factory
from ragas.metrics import (
    Faithfulness,
    ResponseRelevancy,
    LLMContextRecall,
    LLMContextPrecisionWithoutReference
)

# Khởi tạo scorers
evaluator_llm = llm_factory()
evaluator_embeddings = embedding_model
scorers = [
    Faithfulness(llm=evaluator_llm),
    ResponseRelevancy(llm=evaluator_llm, embeddings=evaluator_embeddings),
    LLMContextRecall(llm=evaluator_llm),
    LLMContextPrecisionWithoutReference(llm=evaluator_llm)
]

OUTPUT_FILE = "ragas_eval_results.csv"

async def eval_dataset(dataset, scorers):
    results = []

    # Nếu file chưa có thì tạo header
    columns = ["index", "user_input"] + [scorer.__class__.__name__ for scorer in scorers]
    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        f.write(",".join(columns) + "\n")

    # Duyệt từng sample
    for i, sample in enumerate(dataset.samples):
        row = {"index": i, "user_input": sample.user_input}
        for scorer in scorers:
            try:
                score = await scorer.single_turn_ascore(sample)
            except Exception as e:
                score = None
            row[scorer.__class__.__name__] = score

        results.append(row)

        # Ghi ra file
        df_row = pd.DataFrame([row])
        df_row.to_csv(OUTPUT_FILE, mode="a", header=False, index=False, encoding="utf-8")

        # In ra màn hình
        print(f"\n==== Sample {i} ====")
        print(f"User input: {sample.user_input}")
        for scorer in scorers:
            print(f"{scorer.__class__.__name__}: {row[scorer.__class__.__name__]}")
        print("====================\n")

    return results


# Chạy eval
all_results = asyncio.run(eval_dataset(dataset, scorers))

# Load toàn bộ kết quả cuối để xem nhanh
df = pd.DataFrame(all_results)
print(df.head())
```

```python
import os
import pandas as pd
import numpy as np
from datasets import Dataset, load_dataset
from ragas import evaluate
from ragas.metrics import (
    context_precision,
    context_recall,
    answer_relevancy,
    faithfulness,
)
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.embeddings import HuggingFaceEmbeddings

os.environ["GOOGLE_API_KEY"] = API_KEY

DATA_PATH = "/kaggle/input/realmar/ragas_eval_complete_v2.jsonl"
OUTPUT_CSV = "/kaggle/working/ragas_eval_results.csv"
# ----------------------------

# 1) Load raw dataset
dataset_raw = load_dataset("json", data_files=DATA_PATH, split="train")
df = dataset_raw.to_pandas()
print("Raw dataset rows:", len(df))
print("Columns:", df.columns.tolist())

# 2) Normalization helpers (robust)
def ensure_list_of_strings(x):
    """Trả về list[str] ít nhất một phần tử. Nếu rỗng/None -> ['N/A']"""
    # None / nan
    if x is None:
        return ["N/A"]
    if isinstance(x, float) and pd.isna(x):
        return ["N/A"]
    # numpy array -> tolist
    if isinstance(x, np.ndarray):
        x = x.tolist()
    # if already list: flatten one level and stringify
    if isinstance(x, list):
        flattened = []
        for item in x:
            if item is None:
                continue
            if isinstance(item, np.ndarray):
                item = item.tolist()
            if isinstance(item, list):
                for sub in item:
                    if sub is None:
                        continue
                    flattened.append(str(sub))
            else:
                flattened.append(str(item))
        # strip and remove empty strings
        cleaned = [s.strip() for s in flattened if isinstance(s, str) and s.strip() != ""]
        return cleaned if len(cleaned) > 0 else ["N/A"]
    # if string -> wrap
    if isinstance(x, str):
        s = x.strip()
        return [s] if s else ["N/A"]
    # other scalar types
    return [str(x)]

def ensure_reference_str(x):
    """Trả về chuỗi cho reference (ground-truth answer). Nếu là list -> lấy phần tử phi-rỗng đầu."""
    if x is None:
        return "N/A"
    if isinstance(x, float) and pd.isna(x):
        return "N/A"
    if isinstance(x, np.ndarray):
        x = x.tolist()
    if isinstance(x, list):
        # tìm phần tử chuỗi không rỗng đầu tiên (nếu có)
        for item in x:
            if item is None:
                continue
            if isinstance(item, (list, np.ndarray)):
                # flatten inner
                for sub in (item.tolist() if isinstance(item, np.ndarray) else item):
                    if isinstance(sub, str) and sub.strip():
                        return sub.strip()
            else:
                s = str(item).strip()
                if s:
                    return s
        return "N/A"
    if isinstance(x, str):
        s = x.strip()
        return s if s else "N/A"
    # fallback
    return str(x)

def ensure_text_str(x):
    """Đảm bảo user_input/response là string; nếu rỗng -> 'N/A'"""
    if x is None:
        return "N/A"
    if isinstance(x, float) and pd.isna(x):
        return "N/A"
    if isinstance(x, (list, np.ndarray)):
        # try to extract first non-empty element
        lst = x.tolist() if isinstance(x, np.ndarray) else x
        for it in lst:
            if it is None:
                continue
            s = str(it).strip()
            if s:
                return s
        return "N/A"
    s = str(x).strip()
    return s if s else "N/A"

# 3) Apply normalization to columns we care about
# If your dataset uses different column names, change them here.
for col in ["retrieved_contexts"]:
    if col in df.columns:
        df[col] = df[col].apply(ensure_list_of_strings)
    else:
        df[col] = [["N/A"]] * len(df)

for col in ["reference"]:
    if col in df.columns:
        df[col] = df[col].apply(ensure_reference_str)
    else:
        df[col] = ["N/A"] * len(df)

# Ensure user_input & response are strings
for col in ["user_input", "response"]:
    if col in df.columns:
        df[col] = df[col].apply(ensure_text_str)
    else:
        df[col] = ["N/A"] * len(df)

# Quick stats
num_ref_na = (df["reference"] == "N/A").sum()

def is_ctx_na(lst):
    """Trả về True nếu contexts rỗng hoặc chỉ chứa 'N/A'"""
    if not isinstance(lst, list):
        return True
    if len(lst) == 0:
        return True
    cleaned = [s for s in lst if isinstance(s, str) and s.strip()]
    return all(s == "N/A" for s in cleaned) or len(cleaned) == 0

num_ctx_na = df["retrieved_contexts"].apply(is_ctx_na).sum()

print(f"After normalize: rows={len(df)}, reference=='N/A': {num_ref_na}, contexts 'effectively empty': {num_ctx_na}")


# reset index (avoid keeping old pandas index column)
df = df.reset_index(drop=True)

# 4) convert back to HuggingFace Dataset
dataset = Dataset.from_pandas(df)

# 5) prepare LLM + embeddings
gemini_llm = LangchainLLMWrapper(
    ChatGoogleGenerativeAI(
        model="gemini-2.0-flash",
        temperature=0.0,
        max_output_tokens=512,
    )
)
hf_embeddings = LangchainEmbeddingsWrapper(
HuggingFaceEmbeddings(
        model_name="dangvantuan/vietnamese-document-embedding",
        model_kwargs={"trust_remote_code": True}
    )
)

# 6) Evaluate
# NOTE: set raise_exceptions=False so ragas will not stop at first error.
try:
    result = evaluate(
        dataset=dataset,
        metrics=[context_precision, context_recall, answer_relevancy, faithfulness],
        llm=gemini_llm,
        embeddings=hf_embeddings,
        show_progress=True,
        raise_exceptions=True,   # <--- important to avoid immediate crash
        column_map={
            "question": "user_input",
            "answer": "response",
            "contexts": "retrieved_contexts",
            "ground_truth": "reference"
        }
    )
except Exception as e:
    # fallback: print error and re-raise if you want
    print("evaluate() raised an exception:", repr(e))
    raise

# 7) Print and save results
print("Summary result object:", result)
try:
    df_result = result.to_pandas()
    print("Result DataFrame shape:", df_result.shape)
    # if ragas included an 'error' column, show a little
    if "error" in df_result.columns:
        print("Errors count:", df_result["error"].notna().sum())
    df_result.to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")
    print("Saved detailed results to:", OUTPUT_CSV)
except Exception as e:
    print("Could not convert result to pandas or save:", repr(e))
```
