# pipeline RAG-KG

````python
def build_prompt(question, schema_text):
    return f"""
Bạn là một hệ thống chuyên sinh Cypher query từ câu hỏi tiếng Việt.
Dưới đây là schema liên quan trong knowledge graph:

{schema_text}

Nhiệm vụ:
1. Phân tích câu hỏi: "{question}" để lấy ra các thực thể và mối quan hệ trong câu.
2. Chọn các label và relationship từ schema ở trên.
3. Thêm backticks quanh relationship
4. Sinh ra các cypher có thể có và hợp lệ.
5. Mỗi query phải trả về cả node và edge liên quan, không chỉ giá trị.
6. Lưu ý quan trọng: Lúc nào cũng sinh ra cypher không có label, ví dụ: MATCH (e)-[r]-(l) WHERE toLower(e.name) = toLower("Điện Biên Phủ") RETURN e.name, r, l.name.
Đây là cypher quan trọng nhất nên có lúc nào cũng phải có và đặt lên đầu.
7. Ngoài ra có thể tìm theo thuộc tính `name` gần giống:
    - Dùng `CONTAINS`, `toLower(...)`
8. Thêm toLower() để so khớp cả chữ hoa và chữ thường, ví dụ: MATCH (n)-[r]-(m) WHERE toLower(n.name) = toLower("Cách mạng Tháng Tám") RETURN n, r, m
Output: lưu ý quan trọng là chỉ in ra Cypher query, không chứa các ký tự như ``` cypher ```. Không đính kèm ``` cypher ```
"""

def generate_with_gemini(prompt):
    headers = {"Content-Type": "application/json"}
    payload = {
        "contents": [
            {"parts": [{"text": prompt}]}
        ]
    }
    response = requests.post(API_URL, headers=headers, json=payload)
    data = response.json()
    text = data["candidates"][0]["content"]["parts"][0]["text"]

    return text


def parse_queries_to_list(queries_str: str):
    # Xóa code block markdown ```cypher ... ``` hoặc ``` ... ```
    queries_str = re.sub(r"```[a-zA-Z]*\n?", "", queries_str)
    queries_str = queries_str.replace("```", "")

    queries = []
    current_query = []

    for line in queries_str.strip().splitlines():
        line = line.strip()
        if not line:
            continue

        current_query.append(line)

        # Nếu dòng này có RETURN -> kết thúc 1 query
        if re.search(r"\bRETURN\b", line, re.IGNORECASE):
            queries.append({"query": " ".join(current_query)})
            current_query = []

    # Nếu còn query chưa đóng RETURN thì cũng thêm vào (optional)
    if current_query:
        queries.append({"query": " ".join(current_query)})

    return queries


import re
def fix_cypher_query(query: str) -> str:
    # Sửa lỗi thiếu dấu " đóng chuỗi
    quote_count = query.count('"')
    if quote_count % 2 != 0:
        query += '"'

    # Nếu kết thúc bằng WITH hoặc MATCH mà không có RETURN
    if re.search(r"\bWITH\s+\w+\s*$", query) or re.search(r"\bMATCH\s+\(.*\)\s*WHERE\s+.*$", query):
        query += " RETURN *"

    # Nếu không có RETURN ở cuối, thêm RETURN *
    parts = query.split("RETURN")
    if len(parts) > 2:
        query = "RETURN".join(parts[:2])
    return query


def run_cypher(queries, driver):
    results = []
    max_records = 50

    with driver.session() as session:
        for query in queries:
            text = fix_cypher_query(query["query"])
            # print("\n\nText:", text, "\n",type(text))
            print(f"Đang chạy truy vấn:\n{text}\n")
            try:
                result = session.run(text)
                for i, record in enumerate(result, start=1):
                    results.append(dict(record))
                    if i >= max_records:   # dừng nếu quá nhiều
                        print(f"[CẢNH BÁO] Dừng ở {max_records} bản ghi để tránh OOM.")
                        break
            except Exception as e:
                print("[LỖI] Truy vấn lỗi:")
                print(query["query"])
                print(e)

    return results

def search_schema(query: str, k):
    results_label = schema_label_db.similarity_search(query, k=k)
    results_rel = schema_rel_db.similarity_search(query, k=k)

    schema = "\n".join(
        [f"- {r.page_content}" for r in results_label] +
        [f"- {r.page_content}" for r in results_rel]
    )
    return schema

def generate_cypher(question: str, k):
    # lấy schema liên quan từ vectorDB
    schema_text = search_schema(question, k=k)  # dạng string
    # print(schema_text)
    prompt = build_prompt(question, schema_text)
    return generate_with_gemini(prompt)
````

# Các hàm chuẩn hóa kết quả từ Neo4j

## Đưa kết quả từ Neo4j (KG) thành json list

````python
from neo4j.graph import Node, Relationship

def kg_to_json(results):
    """
    Chuyển kết quả session.run() từ Neo4j thành JSON list (triples).
    - Tự động duyệt các field trong record.
    - Node -> {label, properties}
    - Relationship -> {type, properties}
    """
    json_results = []

    for record in results:
        triple = {}
        for key, value in record.items():
            if isinstance(value, Node):
                triple[key] = {
                    "labels": list(value.labels),
                    "properties": dict(value)
                }
            elif isinstance(value, Relationship):
                triple[key] = {
                    "type": value.type,
                    "properties": dict(value)
                }
            else:
                triple[key] = value  # nếu chỉ là giá trị primitive (str, int, ...)

        json_results.append(triple)

    return json_results

# json_results = kg_to_json(results)
# print(kg_to_json(results))
````

## Đưa kết quả từ json list thành câu văn

````python
def kg_to_natural(json_results):
    """
    Chuyển JSON từ Neo4j (node + relationship) thành danh sách câu tự nhiên tiếng Việt.
    - Nếu có triple chuẩn (source, relationship, target) -> sinh câu trực tiếp.
    - Nếu là kết quả tổng quát (node/rel bất kỳ) -> cố gắng ghép lại.
    """
    sentences = set()

    for triple in json_results:
        # Trường hợp chuẩn source-relationship-target
        if "source" in triple and "relationship" in triple and "target" in triple:
            src = triple["source"]
            rel = triple["relationship"]
            tgt = triple["target"]

            rel_text = rel.replace("_", " ")
            sentence = f"\"{src['name']}\" {rel_text} \"{tgt['name']}\"."
            sentences.add(sentence)
            continue

        # Trường hợp tổng quát hơn (từ kg_to_json)
        nodes = {k: v for k, v in triple.items() if "labels" in v}
        rels = {k: v for k, v in triple.items() if "type" in v}

        for r_key, r_val in rels.items():
            rel_text = r_val["type"].replace("_", " ")
            rel_props = r_val.get("properties", {})

            # Ghép 2 node đầu tiên với quan hệ
            node_keys = list(nodes.keys())
            if len(node_keys) >= 2:
                src = nodes[node_keys[0]]
                tgt = nodes[node_keys[1]]
                # src_name = src["properties"].get("name", "Unknown")
                # tgt_name = tgt["properties"].get("name", "Unknown")
                src_name = src["properties"].get("name")
                tgt_name = tgt["properties"].get("name")

                if not src_name or not tgt_name:  
                    continue
                    
                props_text = ""
                if rel_props:
                    props_text = " (" + ", ".join(f"{k}={v}" for k, v in rel_props.items()) + ")"

                sentence = f"\"{src_name}\" {rel_text} \"{tgt_name}\"{props_text}."
                sentences.add(sentence)

    return sentences


# print(kg_to_natural(json_results))
````

````python
from langchain_core.documents import Document


# Chuyển list Document thành JSON
def kg_to_json_from_documents(documents):
    json_results = []
    for doc in documents:
        entity = {}
        # Lấy tất cả metadata (doc.metadata là dict)
        for key, value in doc.metadata.items():
            entity[key] = value
        # Thêm nội dung
        entity["content"] = doc.page_content
        json_results.append(entity)
    return json_results

# Chuyển JSON thành câu tự nhiên
def kg_to_natural_from_json(json_results):
    sentences = set()
    for item in json_results:
        title = item.get("title") or item.get("name") or "Unknown"
        content = item.get("content", "")
        sentence = f"{title}: {content}"
        sentences.add(sentence)
    return sentences



````

````python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

# model_name = "mistralai/Mistral-7B-Instruct-v0.3"
model_name ="meta-llama/Llama-3.2-3B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_name)

quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=quantization_config,
    device_map="auto",
)

tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=True)


def generate_with_llama(prompt, model):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    
    if "token_type_ids" in inputs:
        del inputs["token_type_ids"]
    
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=1024,
            pad_token_id=tokenizer.eos_token_id,
            eos_token_id=tokenizer.eos_token_id,
            repetition_penalty=1.2,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
        )

    # chỉ lấy phần sinh thêm (không gồm prompt)
    generated = outputs[0][inputs["input_ids"].shape[-1]:]
    text = tokenizer.decode(generated, skip_special_tokens=True)
    return text
````

# Merge các hàm lại thành 1 luồng hoàn chỉnh

````python
def rag_kg_query(model, user_query, k_chunk, k_kg):
    # 1. Sinh Cypher query
    schema_text = search_schema(user_query, k_kg)
    prompt = build_prompt(user_query, schema_text)
    cypher_text = generate_cypher(user_query, 10)
    
    raw_cyphers = parse_queries_to_list(cypher_text)

    if not raw_cyphers:
        return "Không sinh được query"

    print("raw_cyphers--------------")
    print(raw_cyphers)
    print("raw_cyphers--------------")
    
    # 2. Chạy query trên Neo4j
    results = run_cypher(raw_cyphers, driver)
    json_results = kg_to_json(results)

    # 3. Convert thành câu tự nhiên
    natural_sentences = kg_to_natural(json_results)
    print("\nnatural_sentences--------------")
    print(natural_sentences)
    print("natural_sentences--------------")
    # 4. Tạo câu trả lời gọn
    # answer_context = "\n".join(natural_sentences)
    # print(answer_context)

    # 5. Lấy context
    chunk_context = chunk_db.similarity_search(user_query, k_chunk)
    chunk_results = kg_to_json_from_documents(chunk_context)
    chunk_natural_sentences = kg_to_natural_from_json(chunk_results)
    print("\nchunk_natural_sentences--------------")
    print(chunk_natural_sentences)
    print("chunk_natural_sentences--------------\n")

    context = list(natural_sentences) + list(chunk_natural_sentences)
    print("\ncontext-------------------")
    print(context)
    print("context-------------------\n")
    
    final_prompt = f"""
Với câu hỏi của người dùng: {user_query}
Dựa vào dữ liệu trích xuất từ KG:
{natural_sentences}
Dựa vào dữ liệu ngữ cảnh: 
{chunk_natural_sentences}
Yêu cầu:
1. Chỉ sử dụng thông tin trong KG và ngữ cảnh ở trên để trả lời.
2. Nếu không tìm thấy thông tin, hãy trả lời: "Xin lỗi, tôi không tìm thấy thông tin trong dữ liệu đã cung cấp."
3. Kết hợp cả thông tin KG và ngữ cảnh (nếu có) để đưa ra câu trả lời
4. Không sử dụng kiến thức bên ngoài, chỉ đưa ra câu trả lời dựa vào các thông tin mà tôi cung cấp.
Đáp:
"""
    return generate_with_llama(final_prompt, model), context


````
